```python
df.columns = df.columns.str.replace('ï»¿', '')
```

```python
df.columns = [
    col.replace('ï»¿', '').strip().lower()
    for col in df.columns
]
```

```python
df["district"] = df["district"].str.lower()
df["season"] = df["season"].str.lower()
df["region"] = df["region"].str.lower()
df["accomodation category"] = df["accomodation category"].str.lower()
df["accomodation type"] = df["accomodation type"].str.lower()
df["accomodation name"] = df["accomodation name"].str.lower()
df["website"] = df["website"].str.lower()
df["location"] = df["location"].str.lower()
df["location on maps"] = df["location on maps"].str.lower()
df["reservation"] = df["reservation"].str.lower()

```
